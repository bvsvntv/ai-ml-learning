## Basics of memory

```python
from langgraph.store.memory import InMemoryStore
```

```python
# create store
store = InMemoryStore()
```

```python
# create a namespace
namespace = ("user", "u1")
```

```python
# add memories
store.put(namespace, "1", {"data": "User likes exercise."})
store.put(namespace, "2", {"data": "User prefers suits."})
```

```python
# retrieve memories
store.get(namespace, "2")
```

```python
# retrieve all memories from namespace
memories = store.search(namespace)

print('-'*37,"ALL MEMORIES","-"*37)
for memory in memories:
  print(memory.value)
```

```python
namespace2 = ("user", "u2")

store.put(namespace2, "1", {"data": "User is software developer"})
store.put(namespace2, "2", {"data": "User likes blue berry cheese cake."})

memory1 = store.get(namespace2, "1")
print(memory1.value)

memories = store.search(namespace)
print('-'*37,"ALL MEMORIES","-"*37)
for memory in memories:
  print(memory.value)
```

```python
from langchain_openai import OpenAIEmbeddings
from google.colab import userdata
```

```python
api_key = userdata.get("OPENAI_API_KEY")

embedding_model = OpenAIEmbeddings(api_key=api_key, model="text-embedding-3-small")
```

```python
# pass embedding model and define embedding dimensions
semantic_store = InMemoryStore(index={"embed": embedding_model, 'dims': 1536})
```

```python
namespace3 = ("users", "u3")
```

```python
random_memories = [
    "User likes drinking coffee while working in the morning.",
    "User primarily uses Python for software development.",
    "User doesn't like bright themes or white backgrounds in applications.",
    "User prefers CLI tools over graphical user interfaces.",
    "User enjoys traveling to new places and exploring different cultures.",
    "User listens to lo-fi music while coding.",
    "User is currently trying to improve public speaking skills.",
    "User dislikes noisy environments when concentrating on work.",
    "User's favorite food is pizza.",
    "User exercises three times a week.",
    "User uses Vim/Neovim for text editing.",
    "User is interested in artificial intelligence and machine learning.",
    "User recently started learning Rust.",
    "User owns a mechanical keyboard with tactile switches.",
    "User usually works late at night.",
    "User has experience building REST APIs with FastAPI.",
    "User avoids using unnecessary dependencies in projects.",
    "User enjoys reading science fiction novels.",
    "User's preferred operating system is Linux.",
    "User is working on building AI agents using LangGraph."
]

for index, memory in enumerate(random_memories):
  store.put(namespace3, str(index), {"data": memory})
```

```python
retrieved_memory = store.search(namespace3, query="What are user's preferences?", limit=20)
for memory in retrieved_memory:
  print(memory.value)
```

```python
retrieved_memory2 = store.search(namespace3, query="Does the user prefer terminal-based applications?", limit=1)
for memory in retrieved_memory2:
  print(memory.value)
```
